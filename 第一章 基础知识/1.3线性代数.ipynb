```python
#  张量的数据运算和矩阵的数据运算 结构一样
```

```python
x = torch.arange(20)
```

```python
x;x.sum()
```

```python
A = x.reshape(4,5)
```

```python
A_sum_axis0 = A.sum(axis=0)  # 0代表行加
```

```python
A_sum_axis0,A_sum_axis0.shape,A  # ,变成元组了
```

```python
A_sum_axis1 = A.sum(axis=1)  # 1代表列加
```

```python
A_sum_axis1,A_sum_axis1.shape,A 
```

```python
A_sum_axis0
```

```python
C = A.clone().detach()
```

```python
C
```

```python
# 非降维求和
```

```python
sum_A = A.sum(axis=1,keepdim=True);sum_B = A.sum(axis=1)  # keepdim保持原来的维度不变
```

```python
sum_A,sum_B
```

```python
A
```

```python
A/sum_A  # 广播机制 行归一化
```

```python
A.cumsum(axis=0)  # 按行求和
```
