```python
# AlexNet采用暂退法 随机舍去神经元 有时可能增加大量的图像增强数据
import torch
from torch import nn
from d2l import torch as d2l
from torchinfo import summary
```

```python
net = nn.Sequential(
    nn.Conv2d(1,96,kernel_size=11,stride=4,padding=1),nn.ReLU(),  # 输入通道数 输出通道数 核 步长 填充
    nn.MaxPool2d(kernel_size=3, stride=2),  # 汇聚层3x3 步长2
    nn.Conv2d(96,256,kernel_size=5,padding=2),nn.ReLU(),
    nn.MaxPool2d(kernel_size=3, stride=2),
    nn.Conv2d(256,384,kernel_size=3,padding=1),nn.ReLU(),
    nn.Conv2d(384,384,kernel_size=3,padding=1),nn.ReLU(),
    nn.Conv2d(384,256,kernel_size=3,padding=1),nn.ReLU(),
    nn.MaxPool2d(kernel_size=3, stride=2),
    nn.Flatten(),  # 展平
    nn.LazyLinear(4096),nn.ReLU(),  # 全连接层  当我不知道输入参数的时候 设置参数后置
    nn.Dropout(0.5),  # 暂退法 dropout层减轻过拟合
    nn.Linear(4096,4096),nn.ReLU(), nn.Dropout(0.5),
    nn.Linear(4096,10)  # 最终输出
)
```

```python
summary(net,input_size=(1,1,224,224))  # 卷积输入必须四维 (批次, 通道, 高, 宽)
```

```python
batch_size = 128
train_iter, test_iter = d2l.load_data_fashion_mnist(batch_size, resize=224)  # resize 适配深度学习模型的输入尺寸 一般都是224
```

```python
lr, num_epochs = 0.01, 10
d2l.train_ch6(net,train_iter,test_iter,num_epochs,lr,d2l.try_gpu())
```
