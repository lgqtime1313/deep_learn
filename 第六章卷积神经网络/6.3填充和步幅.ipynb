```python
# 卷积核通常是奇数
import torch
from torch import nn
```

```python
def comp_conv2d(conv2d,X):
    X = X.reshape((1,1) + X.shape)
    Y = conv2d(X)
    return Y.reshape(Y.shape[2:])  # 没看懂这里为什么要省略前两个维度
```

```python
conv2d = nn.Conv2d(1,1,kernel_size=3,padding=1)  # 创建卷积核
conv2d
```

```python
X = torch.rand(size=(8,8))
comp_conv2d(conv2d,X).shape
```

```python
X.shape
```

```python
conv2d = nn.Conv2d(1,1,kernel_size=(5,3),padding=(2,1))  # 卷积核为(5，3) 高度和宽度填充(2,1) 这个填充是怎么体现的
comp_conv2d(conv2d,X).shape
```

```python
# 步幅
conv2d = nn.Conv2d(1,1,kernel_size=3,padding=1,stride=2)
comp_conv2d(conv2d,X).shape
```

```python
conv2d = nn.Conv2d(1,1,kernel_size=(3,5),padding=(0,1),stride=(3,4))
comp_conv2d(conv2d,X).shape
```
