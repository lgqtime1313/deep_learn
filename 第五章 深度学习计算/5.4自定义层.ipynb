```python
import torch
from torch import nn
import torch.nn.functional as F
from torchinfo import summary
```

```python
class CenteredLayer(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, X):
        return X - X.mean()
```

```python
layer = CenteredLayer()
M = layer(torch.FloatTensor([1, 2, 3, 4, 5]))
M.mean()
```

```python
net = nn.Sequential(nn.Linear(8,128),CenteredLayer())
```

```python
Y = net(torch.rand(4,8))
Y.mean()
```

```python
print(net)
```

```python
summary(net,input_size=(4,8))
```

```python
class MyLinear(nn.Module):
    def __init__(self,in_units,units):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(in_units,units))
        self.bias = nn.Parameter(torch.randn(units))

    def forward(self, x):
        linear = torch.matmul(x,self.weight.data) + self.bias.data  # matmul矩阵相乘
        return F.relu(linear)
```

```python
linear = MyLinear(5,3)
```

```python
linear.weight
```

```python
linear(torch.rand(2,5))
```

```python
net = nn.Sequential(MyLinear(3,5),CenteredLayer(),MyLinear(5,3))
Y = net(torch.zeros(2,3))
```
