```python
import torch
x = torch.arange(4.0)
x
```

```python
x.requires_grad_(True)  # 等价于 x=torch.arange(4.0,requires_grad=True)
x.grad
```

```python
x
```

```python
y = 2*torch.dot(x,x)
```

```python
y
```

```python
y.backward()  # 计算储存在x.grad中 同时运行一遍计算图被释放
x.grad
```

```python
x.grad == 4*x
```

```python
x.grad.zero_()
```

```python
y.backward()  #  反向传播
```

```python
def f(a):
    b = a * 2
    while b.norm() < 1000:
        b = b * 2
    if b.sum() > 0:
        c = b
    else:
        c = 100 * b
    return c
```

```python
a = torch.randn(size=(),requires_grad=True)
```

```python
a
```

```python
d = f(a)
```

```python
d.backward()
```

```python
a.grad == d / a
```
